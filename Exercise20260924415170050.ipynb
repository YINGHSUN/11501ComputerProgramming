## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？ 因為 Python 的 int 採用的是 任意精度整數（Arbitrary Precision Integer） 設計，而不是直接使用 CPU 的固定長度整數型別。
2. Python 的 `int` 是否具有固定的位元數限制？ 沒有固定限制
3. 什麼是 Arbitrary Precision（任意精度）？ 整數的位數不是在設計時固定死，而是根據需要動態成長。
4. CPython 使用什麼物件來實作 `int`？ PyLongObject
5. Python 整數大小真正的限制是什麼？ 理論限制

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？ 因為 Python 的 float 採用 IEEE 754 雙精度浮點數（double precision floating-point） 格式儲存數值，而許多十進位小數無法用有限個二進位位元精確表示。
2. 浮點數是否能精確表示所有小數？ 不能 只有某些特殊小數可以被精確表示。
3. 什麼是浮點數的精度（precision）？ 可以可靠保存多少位有效數字。
4. 什麼是浮點數的範圍（range）？ 能表示多大或多小的數值。
5. 精度與範圍有何不同？ 範圍：能到多大、多小 精度：能分辨多細

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int | 不固定（任意精度） | 無誤差（整數可精確表示） | 計數、索引、密碼學、大整數運算|
| FP64 | 64 bits | 有誤差 | 科學計算、工程模擬、高精度數值分析 |
| FP32 | 32 bits | 有誤差 | AI 訓練、圖形處理、一般數值運算 |
| FP16 | 16 bits | 有誤差（較大） | AI 推論、GPU 加速、節省記憶體 |
| BF16 | 16 bits | 有誤差（較大） | 大型 AI 模型訓練與推論 |

並回答：

1. 哪種表示方式最精確？ int 因為int 可以精確表示整數，沒有四捨五入誤差。
2. 哪種表示方式最節省記憶體？ FP16 與 BF16 最省記憶體。
3. 哪種表示方式最常見於 AI 訓練？ BF16 因為記憶體需求只有 FP32 的一半、範圍接近 FP32、不overflow/underflo、訓練穩定
4. 哪種表示方式最常見於 AI 推論（Inference）？ FP16 推論重視速度、推論重視記憶體、推論重視部署成本

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？ 整數的位數不是事先固定為 8、16、32 或 64 位元，而是可以根據需要動態增加。 
2. 它是否代表可以儲存無限大的整數？ 不能 RAM 有限、硬碟有限、位址空間有限、時間有限
3. CPython 如何實現這項特性？ 使用PyLongObject來實作 Python 的 int。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？ 因為 CPython 的 int 並不是直接使用 CPU 的 32-bit 或 64-bit 整數。
2. 為什麼 `float` 卻存在範圍限制？ float 的範圍來自固定的 64-bit 結構，而不是記憶體大小。
3. 為什麼 `float` 可能產生誤差？ 因為大部分十進位小數無法被有限長度的二進位精確表示。

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。 Python 的 int 追求：「數學上正確的結果最重要」 而AI是「結果接近即可，不一定要完全精確」
2. 各自的優點與缺點。 python 優點：完全精確、不會因位數增加而溢位、適合金融、密碼學、數學運算
                            缺點：記憶體需求大、運算速度較慢
                    FP16 / BF16  優點： 記憶體小、GPU 運算極快、可訓練更大的模型
                                 缺點： 存在誤差、 不適合需要完全精確的運算  
3. 為什麼大型語言模型通常選擇較低精度的表示方式？ 因為大型語言模型最缺的不是精度，而是：1.GPU記憶體2.頻寬3.運算量4.能源
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？ 問題一：記憶體暴增  問題二：訓練速度下降  問題三：電力消耗增加  問題四：GPU 利用率下降  問題五：精度提升有限
