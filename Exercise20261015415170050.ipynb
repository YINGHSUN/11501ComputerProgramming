習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

import random

r = random.random()      # 產生 0 到 1 之間的小數
x = r                    # 保留原值供計算使用

binary = ""

count = 0
while count < 20:
    x = x * 2

    if x >= 1:
        binary += "1"
        x = x - 1
    else:
        binary += "0"

    count += 1

print(f"r = {r}")
print(f"r ≈ 0.{binary}_2")

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`
- `sys.getsizeof()`
- `PyLongObject`

三者之間的關係。

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。

n = input().strip()

num_dict = {
    '0': '零', '1': '壹', '2': '貳', '3': '參', '4': '肆',
    '5': '伍', '6': '陸', '7': '柒', '8': '捌', '9': '玖'
}

small_unit = ['', '拾', '佰', '仟']

big_unit = [
    '', '萬', '億', '兆', '京', '垓', '秭', '穰', '溝',
    '澗', '正', '載', '極', '恆河沙', '阿僧祇',
    '那由他', '不可思議', '無量大數'
]


def convert_four(s):
    result = ''
    zero_flag = False
    length = len(s)

    for i, ch in enumerate(s):
        digit = int(ch)
        pos = length - 1 - i

        if digit == 0:
            zero_flag = True
        else:
            if zero_flag and result:
                result += '零'
            result += num_dict[ch] + small_unit[pos]
            zero_flag = False

    return result


# 每四位分組
groups = []
while n:
    groups.append(n[-4:])
    n = n[:-4]

answer = ''

for i in range(len(groups) - 1, -1, -1):
    part = convert_four(groups[i])

    if part:
        answer += part + big_unit[i]

        # 處理中間缺位的零
        if i > 0:
            next_group = groups[i - 1]
            if int(next_group) != 0 and len(next_group) < 4:
                answer += '零'

print(answer)

習題三：請勿使用for、range，而是利用iterable、iter()、next()、while、StopIteration來設計九九乘法表。

nums = (1, 2, 3, 4, 5, 6, 7, 8, 9)

outer = iter(nums)

while True:
    try:
        i = next(outer)

        inner = iter(nums)

        while True:
            try:
                j = next(inner)
                print(f"{i}*{j}={i*j}", end="\t")
            except StopIteration:
                break

        print()

    except StopIteration:
        break
